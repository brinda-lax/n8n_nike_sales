```python
null_columns = df.columns[df.isna().any()].tolist()

print(null_columns)
```

```python
for col in null_columns:
    if df[col].isna().any():
        df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df['Units_Sold_Flag'] = np.select(
    [
        df['Units_Sold'] < 0,
        df['Units_Sold'] == 0
    ],
    [
        'Negative Quantity',
        'Zero Quantity'
    ],
    default='Valid Quantity'
)
```

```python
df['Units_Sold_Flag'].value_counts()
```

```python
df.loc[df['Revenue'] < 0,
       ['Order_ID', 'Units_Sold', 'MRP',
        'Discount_Applied', 'Revenue', 'Profit']]
```

```python
df['Expected_Revenue'] = (
    df['Units_Sold'] * df['MRP'] *
    (1 - df['Discount_Applied'])
)
```

```python
import numpy as np

mismatch = ~np.isclose(
    df['Revenue'],
    df['Expected_Revenue'],
    atol=0.01
)

mismatch.sum()
```

```python
df.loc[mismatch, [
    'Order_ID',
    'Units_Sold',
    'MRP',
    'Discount_Applied',
    'Revenue',
    'Expected_Revenue',
    'Profit'
]]
```

```python
df['Revenue_Validation'] = np.where(
    np.isclose(df['Revenue'], df['Expected_Revenue'], atol=0.01),
    'Valid Revenue',
    'Revenue Mismatch'
)
```
